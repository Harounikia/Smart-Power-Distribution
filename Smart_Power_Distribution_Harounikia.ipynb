```python

# ۲. تست و اجرای مدل توزیع اقتصادی بار (LP)
import pandas as pd
from lp_solver import EconomicDispatchSolver

print("=== اجرای مدل بهینه‌سازی LP ===")
solver_lp = EconomicDispatchSolver(demand=220, solar_availability=45, battery_capacity=30)
thermal_units = [
    {"name": "حرارتی 1 (پایه)", "min": 20, "max": 150, "cost_per_mw": 40.0},
    {"name": "حرارتی 2 (میان‌باری)", "min": 10, "max": 100, "cost_per_mw": 65.0},
    {"name": "حرارتی 3 (پیک)", "min": 0, "max": 60, "cost_per_mw": 90.0},
]

lp_res = solver_lp.solve(thermal_units)
print(f"وضعیت پاسخ: {lp_res['status']}")
print(f"هزینه کل تولید: {lp_res['total_cost']} $")
print(f"قیمت سایه‌ای (Shadow Price): {lp_res['shadow_price']} $/MW")
print("تخصیص توان واحدها:", lp_res['dispatch'])
```

```python

# ۳. تست و اجرای الگوریتم تکاملی ژنتیک (GA)
import numpy as np
from ga_solver import EVStationGeneticSolver

print("\n=== اجرای الگوریتم ژنتیک GA ===")
np.random.seed(42)
traffic_nodes = np.random.randint(10, 80, size=10)

ga_solver = EVStationGeneticSolver(num_candidate_nodes=10, pop_size=40, generations=80, mutation_rate=0.05)
best_sol, best_fit, history = ga_solver.solve(traffic_nodes)

print(f"چیدمان بهینه شارژرها در گره‌ها: {best_sol}")
print(f"مقدار برازندگی بهینه (Fitness): {best_fit}")
```
